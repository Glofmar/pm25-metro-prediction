# PM2.5 in the Seoul metro: scores and map

## Scores

Simulated campaign on the Seoul data: half of the stations are visited, three times each (about 380 measurements), and the model learns from these visits only. Averages over 10 draws, in µg/m³.

| Tested on | MAE | RMSE | R² | Mean of all visits (MAE) |
|---|---|---|---|---|
| Visited stations, hours not measured | 6.1 | 9.9 | 0.78 | 14.1 |
| Stations never visited | 10.2 | 16.0 | 0.34 | 13.7 |

The error depends on how many stations are visited and how many times. More visits per station improve the visited stations; more stations visited improve the others. Details in [docs/results.md](docs/results.md).
